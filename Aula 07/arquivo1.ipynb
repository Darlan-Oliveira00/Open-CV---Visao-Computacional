```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

import cv2
```

```python
nova_imagem = np.zeros(shape=(1024,1024,3), dtype = np.int16)
```

```python
type(nova_imagem)
```

```python
nova_imagem.shape
```

```python
nova_imagem
```

```python
plt.imshow(nova_imagem)
```

```python
nova_imagem[:,:,:] = 0
```

```python
cv2.rectangle(nova_imagem, pt1 = (200,250), pt2 = (400,450), color = (255, 255, 255), thickness = 5)
```

```python
cv2.rectangle(nova_imagem, pt1 = (400,650), pt2 = (800,950), color = (255, 255, 0), thickness = 10)
```

```python
cv2.circle(nova_imagem, center = (500,500), radius = (75), color = (0, 255, 255), thickness = 10)
```

```python
plt.imshow(nova_imagem)
```

```python
cv2.circle(nova_imagem, center = (500,500), radius = (75), color = (0, 255, 255), thickness = -10)
plt.imshow(nova_imagem)
```

```python
cv2.circle(nova_imagem, center = (300,450), radius = (75), color = (255, 255, 255), thickness = 10)
plt.imshow(nova_imagem)
```

```python
cv2.line(nova_imagem, pt1 = (200,200), pt2 = (870,600), color = (0, 150, 0), thickness = 10)
```

```python
fonte = cv2.FONT_HERSHEY_SIMPLEX
```

```python
cv2.putText(nova_imagem, text = 'Visão Computacional', org = (20,180), fontFace = fonte, fontScale = 2, color = (255,0,0), thickness = 5, lineType = cv2.LINE_AA)
```
