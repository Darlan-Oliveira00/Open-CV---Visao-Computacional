```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def apresentaImg(imagem):
    img = plt.figure(figsize = (30,28))
    ax = img.add_subplot(111)
    ax.imshow(imagem, cmap = 'gray')
```

```python
vinho = cv2.imread('vinho.jpg')
vinho = cv2.cvtColor(vinho, cv2.COLOR_BGR2GRAY)
apresentaImg(vinho)
```

```python
pratileira_vinhos = cv2.imread('pratileira_vinho.jpg')
pratileira_vinhos = cv2.cvtColor(pratileira_vinhos, cv2.COLOR_BGR2GRAY)
apresentaImg(pratileira_vinhos)
```

```python
orb = cv2.ORB_create()
```

```python
kp1, des1 = orb.detectAndCompute(vinho, None)
kp2, des2 = orb.detectAndCompute(pratileira_vinhos,None)
```

```python
forcaBruta = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck = True)
```

```python
matches = forcaBruta.match(des1, des2)
```

```python
matches = sorted(matches, key = lambda x:x.distance)
```

```python
vinho_matches = cv2.drawMatches(vinho, kp1, pratileira_vinhos, kp2, matches[:100], None, flags = 2)
```

```python
apresentaImg(vinho_matches)
```
